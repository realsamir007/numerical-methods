### Bisection Method

```python
def f(x):
    return np.sin(x)-2*x+1
    
def bisection(xl, xu, es, max_iter):

    if f(xl)*f(xu) > 0:
        print("The root does not exists here")
        return None

    iter_count = 0
    xm_old = xl
    ea = 100
    print(f"{'Iter':>4} {'x1':>10} {'xu':>10} {'xm':>10} {'f(xm)':>10} {'ea(%)':>10})")

    while (ea > es) and (iter_count < max_iter):

        xm = (xl + xu)/2

        f_xm = f(xm)
        f_xl = f(xl)

        if iter_count > 0:
            ea = abs((xm-xm_old)/xm)*100
        xm_old = xm

        print(f'{iter_count+1:>4} {xl:>10.6f} {xu:>10.6f} {xm:>10.6f} {f_xm:>10.6f} {ea:>10.6f}')

        if f_xl*f_xm <0:
            xu = xm
        elif f_xl*f_xm >0:
            xl = xm
        else:
            ea = 0

        iter_count +=1

    print(f"Approx Root = {xm:.6f} after {iter_count}: iterations")
    return xm

print("f(x)=sin(x)-2x+1")

root=bisection(xl=0, xu=2, es=0.01, max_iter=100)

```

#### Newton Raphson Method 

```python
import numpy as np
import sympy as sp

# Define the function f(x)
def f(x):
    return sp.exp(-0.5*x) * (4 - x) - 2

# Define the derivative df(x) using sympy
def df(x_val):
    X = sp.Symbol('X')
    f_expr = sp.exp(-0.5*X) * (4 - X) - 2
    dff = sp.diff(f_expr, X)
    return float(dff.evalf(subs={X: x_val}))

# Newton-Raphson method
def newton_method(xi, es, max_iter):
    iter_count = 0
    ea = 100
    print(f"{'Iter':>4} {'xi':>12} {'f(xi)':>12} {'df(xi)':>12} {'x_new':>12} {'ea (%)':>10}")
    
    while ea > es and iter_count < max_iter:
        f_val = f(xi)
        df_val = df(xi)
        
        if df_val == 0:
            print("Zero derivative. No solution found.")
            return None
        
        x_new = xi - (f_val / df_val)
        
        if iter_count > 0:
            ea = abs((x_new - xi) / x_new) * 100
        
        print(f"{iter_count+1:>4} {xi:>12.6f} {f_val:>12.6f} {df_val:>12.6f} {x_new:>12.6f} {ea:>10.4f}")
        
        xi = x_new
        iter_count += 1

    print(f"\nApproximate root = {xi:.6f} after {iter_count} iterations")
    return xi
# Run the Newton Raphson Method
print("f(x) = e^-0.5x(4-x)-2")
# Run the method
root = newton_method(xi=1, es=0.01, max_iter=100)
```

## Secant Method

```python
import numpy as np

# Define f(x) numerically
def f(x):
    return x**2 + sp.cos(x) - sp.exp(-x) - 2

# Secant Method Implementation
def secant_method(x0, x1, es, max_iter):
    iter_count = 0
    ea = 100
    print(f"{'Iter':>4} {'x0':>12} {'x1':>12} {'f(x0)':>12} {'f(x1)':>12} {'x_new':>12} {'ea (%)':>10}")
    
    while ea > es and iter_count < max_iter:
        f_x0 = f(x0)
        f_x1 = f(x1)
        
        if f_x1 - f_x0 == 0:
            print("Zero denominator. No solution found.")
            return None
        
        x_new = x1 - f_x1 * (x1 - x0) / (f_x1 - f_x0)
        
        if iter_count > 0:
            ea = abs((x_new - x1) / x_new) * 100

        print(f"{iter_count+1:>4} {x0:>12.6f} {x1:>12.6f} {f_x0:>12.6f} {f_x1:>12.6f} {x_new:>12.6f} {ea:>10.4f}")
        
        # Prepare for next iteration
        x0 = x1
        x1 = x_new
        iter_count += 1

    print(f"\nApproximate root = {x_new:.6f} after {iter_count} iterations")
    return x_new

# Run the Secant Method
print("f(x) = x^2 + cos(x) - e^-x - 2")
root = secant_method(x0=0.5, x1=1.0, es=0.01, max_iter=100)
```
