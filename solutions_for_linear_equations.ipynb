```python
import numpy as np
```

## 1. Gauss Elimination Method

```python
# Augmented matrix [A | b]
A = np.array([
    [2, -6, -1, -38],
    [-3, -1, 7, -34],
    [-8, -1, -2, -20]
], dtype=float)

n = len(A)

# Forward elimination
for i in range(n):
    # Pivot: make A[i][i] = 1
    A[i] = A[i] / A[i][i]
    
    # Eliminate column below the pivot
    for j in range(i+1, n):
        A[j] = A[j] - A[j][i] * A[i]

# Back substitution
x = np.zeros(n)
for i in range(n-1, -1, -1):
    x[i] = A[i][-1] - np.dot(A[i][i+1:n], x[i+1:n])

# Print the solution
for i, val in enumerate(x, start=1):
    print(f"x{i} = {val:.2f}")
```

## 2. Gauss Jordan Method

```python
# Augmented matrix for the system
A = np.array([
    [2, 3, -1, 54],
    [1, 1,  1, 6],
    [-2, 5, 2, -3]
], dtype=float)

# Number of equations
n = len(A)

# Gauss-Jordan Elimination
for i in range(n):
    # Make the diagonal element 1
    A[i] = A[i] / A[i][i]
    
    # Make the other elements in column i zero
    for j in range(n):
        if i != j:
            A[j] = A[j] - A[j][i] * A[i]

# Extract the solution
solution = A[:, -1]

# Print the result
variables = ['x', 'y', 'z']
for var, val in zip(variables, solution):
    print(f"{var} = {val:.2f}")
```

## 3. Gauss Seidel Method

```python
def gauss_seidel(a, b, x_init, iterations=25, tol=1e-6):
    n = len(a)
    x = x_init.copy()

    print(f"{'Iter':>4} {'x':>12} {'y':>12} {'z':>12}   {'Error x':>12} {'Error y':>12} {'Error z':>12}")
    for it in range(1, iterations + 1):
        x_old = x.copy()

        for i in range(n):
            sum_ = sum(a[i][j] * x[j] for j in range(n) if j != i)
            x[i] = (b[i] - sum_) / a[i][i]

        # Compute absolute errors
        errors = [abs(x[i] - x_old[i]) for i in range(n)]

        # Display values and errors
        print(f"{it:4d} {x[0]:12.6f} {x[1]:12.6f} {x[2]:12.6f}   {errors[0]:12.6f} {errors[1]:12.6f} {errors[2]:12.6f}")

        # Stop if all errors are below tolerance
        if all(error < tol for error in errors):
            print("Converged.")
            break

    return x

# Coefficient matrix A and RHS vector b
A = [
    [6, -2, 1],
    [5, 10, 1],
    [-3, 1, 15]
    
]

b = [15, 28, 44]

# Initial guess for x, y, z
x_init = [1, 1, 1]

# Solve using Gauss-Seidel method
solution = gauss_seidel(A, b, x_init)

# Final result
print("\nFinal Solution:")
variables = ['x', 'y', 'z']
for var, val in zip(variables, solution):
    print(f"{var} = {val:.6f}")
```
