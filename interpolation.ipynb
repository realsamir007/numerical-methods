## 1. Linear Interpolation

```python
import numpy as np

def linearint(xs, ys, x):
    result = ys[0]+(x-xs[0])*(ys[1]-ys[0])/(xs[1]-xs[0])
    return result

x_values = np.array([1,4])
y_values = np.array([5,12])
x_value = 4

ans = linearint(x_values, y_values, x_value)
print("Interpolated value for the given x is:", ans)
```

### 2. Lagrange Interpolation Polynomial

```python
import numpy as np

def lagrange_int(xs, x, y):
    l0_x = (xs - x[1]) * (xs - x[2]) / ((x[0] - x[1]) * (x[0] - x[2]))
    l1_x = (xs - x[0]) * (xs - x[2]) / ((x[1] - x[0]) * (x[1] - x[2]))
    l2_x = (xs - x[0]) * (xs - x[1]) / ((x[2] - x[0]) * (x[2] - x[1]))
    print(f"l0_x: {l0_x}, l1_x: {l1_x}, l2_x: {l2_x}")

    final_ans = y[0] * l0_x + y[1] * l1_x + y[2] * l2_x
    print("The interpolation of the value is:", final_ans)
    return final_ans

# Input data

x_values = np.array([2, 4, 6, 8, 10, 12], dtype=float)
y_values= np.array([30, 29, 27, 25, 25, 23], dtype=float)

x_value = 2.5

# Find closest index and safely pick 3 points
idx = np.searchsorted(x_values, x_value)
# Edge handling
if idx <= 1:
    x_find = x_values[0:3]
    y_find = y_values[0:3]
elif idx >= len(x_values) - 1:
    x_find = x_values[-3:]
    y_find = y_values[-3:]
else:
    x_find = x_values[idx - 1:idx + 2]
    y_find = y_values[idx - 1:idx + 2]

print("Selected x points:", x_find)
print("Selected y points:", y_find)

# Interpolation
ans = lagrange_int(x_value, x_find, y_find)
```

### 3. Newton Interpolation

```python
# Function to calculate divided differences
def divided_diff(x, y):
    n = len(x)
    coef = np.copy(y)
    for j in range(1, n):
        for i in range(n - 1, j - 1, -1):
            coef[i] = (coef[i] - coef[i - 1]) / (x[i] - x[i - j])
    return coef

# Function to evaluate Newton Polynomial at a given x
def newton_poly(coef, x_data, x):
    n = len(coef)
    result = coef[-1]
    for i in range(n - 2, -1, -1):
        result = result * (x - x_data[i]) + coef[i]
    return result

# Example Data
x = np.array([2, 4, 6, 8, 10, 12], dtype=float)
y = np.array([30, 29, 27, 25, 25, 23], dtype=float)

# Compute coefficients
coefficients = divided_diff(x, y)

# Display polynomial coefficients
print("Divided difference coefficients:")
for i, c in enumerate(coefficients):
    print(f"a{i} = {c:.6f}")

# Interpolate at a specific value
x_val = 5
y_interp = newton_poly(coefficients, x, x_val)
print(f"\nInterpolated value at x = {x_val}: {y_interp:.6f}")
```
